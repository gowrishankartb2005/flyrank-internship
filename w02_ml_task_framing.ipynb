## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd
import numpy as np

# Attempt path resolution for starter data
import os
paths = ['../../data/starter.csv', '../data/starter.csv', 'data/starter.csv', '/content/data/starter.csv']
df = None
for p in paths:
    if os.path.exists(p):
        df = pd.read_csv(p)
        break

# Fallback synthetic slice if local file is missing
if df is None:
    df = pd.DataFrame({
        'url': [f'https://example.com/page-{i}' for i in range(100)],
        'impressions_90d': np.random.randint(0, 1500, size=100),
        'avg_position': np.random.uniform(5.0, 45.0, size=100),
        'ctr': np.random.uniform(0.005, 0.12, size=100),
        'trend_direction': np.random.choice(['declining', 'stable', 'growing'], size=100)
    })

# Unit of analysis: 1 row = 1 URL
# Target definition sketch
df['target_recoverable'] = (
    (df['trend_direction'] == 'declining') &
    (df['impressions_90d'] > df['impressions_90d'].median()) &
    (df['avg_position'] < 30)
).astype(int)

print(f"Unit of analysis: URL level (Total rows: {len(df)})")
print("\nFeature slice preview with constructed target:")
display(df[['impressions_90d', 'avg_position', 'ctr', 'trend_direction', 'target_recoverable']].head())
```
